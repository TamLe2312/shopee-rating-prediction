```python
import re
from vietnam_number import n2w

# ============================================================
# HÀM MAPPING SỐ -> CHỮ
# Chỉ nên dùng text to speech
# Không nên dùng cho bài toán phân tích vì nó mất định lượng
# ============================================================
def convert_standalone_numbers(text):
    # Biểu thức chính quy tìm các chữ số đứng một mình (được bao quanh bởi ranh giới từ \b)
    pattern = r'\b\d+\b'
    
    def replace_match(match):
        num_str = match.group(0)
        # Chỉ chuyển đổi nếu là số đơn lẻ (ví dụ từ 0 đến 9 hoặc theo ý muốn của bạn)
        # Nếu muốn áp dụng cho mọi số đứng độc lập, bỏ qua điều kiện len()
        try:
            return n2w(num_str)
        except:
            return num_str

    # Thay thế số bằng chữ tiếng Việt trong toàn bộ chuỗi
    return re.sub(pattern, replace_match, text)
```
