```python
import numpy as np
import pandas as pd
import re

pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_columns", None)

from shopee_rating_preprocessing.utils.present import show_df
```

```python
df = pd.read_csv("../data/processed/02_emoji_mapped.csv")

show_df(df["Comment"].head(10))
```

```python
slang_dict = pd.read_csv('../data/dict/slang_dict.csv')
slang_dict = dict(zip(slang_dict['slang'], slang_dict['meaning']))

# print(slang_dict.items())
```

```python
# Dùng vòng lặp quét qua từng cặp key-value
for old_word, new_word in slang_dict.items():
    # Dùng \b để đánh dấu ranh giới từ, tránh thay nhầm từ bên trong từ khác
    df["Comment"] = df["Comment"].str.replace(r'\b' + re.escape(old_word) + r'\b', new_word, regex=True)
```

```python
show_df(df.loc[:, "Comment"].head())
```

```python
df.to_csv('../data/processed/(3)_slang_mapped.csv', index=False, encoding='utf-8-sig')
```
